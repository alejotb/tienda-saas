# Depuración de sintaxis Dart para `checkout_full_page_widget.dart`

Este notebook analiza el archivo con errores de sintaxis de Flutter/Dart y ayuda a corregir el widget tree final usando el archivo de contexto `ennegram_context` como guía.

```python
from pathlib import Path

path = Path('lib/pages/checkout_full_page/checkout_full_page_widget.dart')
text = path.read_text(encoding='utf-8')
lines = text.splitlines()

paren = 0
brack = 0
brace = 0
max_paren = (0, 0)
max_brack = (0, 0)
max_brace = (0, 0)
for i, line in enumerate(lines, start=1):
    for ch in line:
        if ch == '(':
            paren += 1
        elif ch == ')':
            paren -= 1
        elif ch == '[':
            brack += 1
        elif ch == ']':
            brack -= 1
        elif ch == '{':
            brace += 1
        elif ch == '}':
            brace -= 1
    if paren > max_paren[0]:
        max_paren = (paren, i)
    if brack > max_brack[0]:
        max_brack = (brack, i)
    if brace > max_brace[0]:
        max_brace = (brace, i)
    if brack < 0 or paren < 0 or brace < 0:
        print(f'negative at line {i}: paren={paren} brack={brack} brace={brace}')

print(f'final counts: paren={paren}, brack={brack}, brace={brace}')
print(f'max open paren {max_paren}, max open brack {max_brack}, max open brace {max_brace}')

# show last 20 lines with counts

paren = brack = brace = 0
print('\nCounts at each line from 2200 onwards:')
for i, line in enumerate(lines[2199:], start=2200):
    for ch in line:
        if ch == '(':
            paren += 1
        elif ch == ')':
            paren -= 1
        elif ch == '[':
            brack += 1
        elif ch == ']':
            brack -= 1
        elif ch == '{':
            brace += 1
        elif ch == '}':
            brace -= 1
    print(f'{i}: paren={paren} brack={brack} brace={brace} | {line}')
```

```python
from pathlib import Path

path = Path('lib/pages/checkout_full_page/checkout_full_page_widget.dart')
copy_path = Path('lib/pages/checkout_full_page_copy/checkout_full_page_copy_widget.dart')
text = path.read_text(encoding='utf-8')
copy_text = copy_path.read_text(encoding='utf-8')
start_marker = 'return GestureDetector('
start_idx = text.find(start_marker)
copy_start = copy_text.find(start_marker)
if start_idx == -1 or copy_start == -1:
    raise RuntimeError('start marker not found')
new_text = text[:start_idx] + copy_text[copy_start:]
path.write_text(new_text, encoding='utf-8')
print('Replaced build method from start marker to EOF with copy reference')
```
